```python
import requests
import json
import time
import random
from tqdm import tqdm   # tqdm để có thanh tiến trình đẹp

def crawl_all_reviews(product_id=75307228, spid=75307229, limit=10):
    url = "https://tiki.vn/api/v2/reviews"
    all_reviews = []
    page = 1
    max_page = 300  # an toàn, thực tế sẽ dừng khi hết

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/134.0.0.0 Safari/537.36",
        "Accept": "application/json",
        "Referer": f"https://tiki.vn/sach-osho-yeu-being-in-love-yeu-trong-tinh-thuc-p{product_id}.html"
    }

    print(f"Bắt đầu cào đánh giá sản phẩm {product_id}...")

    while True:
        params = {
            "product_id": product_id,
            "spid": spid,
            "limit": limit,
            "page": page,
            "sort": "id|desc",           # mới nhất trước
            "include": "comments,contribute_info"
        }

        try:
            resp = requests.get(url, params=params, headers=headers, timeout=15)

            if resp.status_code != 200:
                print(f"❌ Lỗi {resp.status_code} tại trang {page}")
                break

            data = resp.json()
            reviews = data.get("data", [])

            if not reviews:
                print("✅ Đã lấy hết tất cả đánh giá!")
                break

            print(f"Trang {page}: lấy được {len(reviews)} đánh giá")

            for review in reviews:
                review_dict = {
                    "review_id": review.get("id"),
                    "title": review.get("title"),
                    "content": review.get("content"),
                    "rating": review.get("rating"),
                    "thank_count": review.get("thank_count"),
                    "created_at": review.get("created_at"),
                    "created_by_name": review.get("created_by", {}).get("name"),
                    "created_by_fullname": review.get("created_by", {}).get("full_name"),
                    "images": [img.get("full_path") for img in review.get("images", [])],
                    "used_time": review.get("timeline", {}).get("content"),
                    "comments": []
                }

                # Lấy bình luận (comment) của seller hoặc khách
                for cmt in review.get("comments", []):
                    review_dict["comments"].append({
                        "commentator": cmt.get("commentator"),   # seller hoặc customer
                        "fullname": cmt.get("fullname"),
                        "content": cmt.get("content"),
                        "create_at": cmt.get("create_at")
                    })

                all_reviews.append(review_dict)

            # Kiểm tra hết trang chưa
            paging = data.get("paging", {})
            if page >= paging.get("last_page", 1):
                print(" Đã cào xong tất cả đánh giá!")
                break

            page += 1
            time.sleep(random.uniform(1.8, 3.8))   # delay an toàn

        except Exception as e:
            print(f"Lỗi tại trang {page}: {e}")
            break

    # Lưu file
    filename = f"tiki_reviews_full_{product_id}.json"
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(all_reviews, f, ensure_ascii=False, indent=4)

    print(f"\n Hoàn thành! Tổng cộng: {len(all_reviews)} đánh giá đã được lưu vào file:")
    print(f"   → {filename}")
    return all_reviews


# ==================== CHẠY Ở ĐÂY ====================
if __name__ == "__main__":
    crawl_all_reviews()
```

```python
import requests
import json
import time
import random

def crawl_all_reviews(product_id, spid, limit=20):
    url = "https://tiki.vn/api/v2/reviews"

    all_reviews = []
    page = 1

    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept": "application/json",
        "Referer": f"https://tiki.vn/p{product_id}.html"
    }

    while True:
        params = {
            "product_id": product_id,
            "spid": spid,
            "limit": limit,
            "page": page,
            "sort": "id|desc",
            "include": "comments,contribute_info"
        }

        try:
            resp = requests.get(url, params=params, headers=headers, timeout=15)

            print("Đang request:")
            print(resp.url)

            if resp.status_code != 200:
                print("Lỗi:", resp.status_code)
                break

            data = resp.json()

            reviews = data.get("data", [])

            if not reviews:
                print("Đã hết review")
                break

            print(f"Trang {page}: {len(reviews)} reviews")

            for review in reviews:
                item = {
                    "review_id": review.get("id"),
                    "rating": review.get("rating"),
                    "content": review.get("content"),
                    "title": review.get("title"),
                    "author": review.get("created_by", {}).get("full_name"),
                    "created_at": review.get("created_at"),
                    "images": [
                        img.get("full_path")
                        for img in review.get("images", [])
                    ],
                    "comments": []
                }

                # Bình luận phản hồi
                for cmt in review.get("comments", []):
                    item["comments"].append({
                        "commentator": cmt.get("commentator"),
                        "fullname": cmt.get("fullname"),
                        "content": cmt.get("content")
                    })

                all_reviews.append(item)

            page += 1

            time.sleep(random.uniform(1, 3))

        except Exception as e:
            print("Lỗi:", e)
            break

    # lưu file
    filename = f"tiki_reviews_{product_id}.json"

    with open(filename, "w", encoding="utf-8") as f:
        json.dump(all_reviews, f, ensure_ascii=False, indent=4)

    print(f"\nĐã lưu {len(all_reviews)} reviews vào {filename}")

    return all_reviews


# ================= RUN =================
crawl_all_reviews(
    product_id=75307228,
    spid=75307229
)
```
